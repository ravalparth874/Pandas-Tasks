### 🟠 Level 3 — Filtering & Conditional Logic

#### Q11. Find all customers from **Surat**.

#### Q12. Find all customers who spent more than **₹20,000** (`TotalSpend > 20000`).

```python
# Write your code here
Spent20 = df[df["TotalSpend"] > 20000]
Spent20
```

#### Q13. Find customers who are from **Surat** AND use **UPI** as their preferred payment method.

```python
# Write your code here
SuratUPI = df[(df["City"] == "Surat") & (df["PreferredPayment"] == "UPI")]
SuratUPI
```

#### Q14. Find customers whose `Age` is between **25 and 35** (inclusive).

#### Q15. Find customers who used a discount (`DiscountUsed == 'Yes'`) OR whose `TotalSpend` is greater than **₹50,000**.

```python
# Write your code here
Customers = df[(df["DiscountUsed"] == "Yes") | (df["TotalSpend"] > 50000)]
Customers
```

### 🟣 Level 5 — GroupBy & Advanced Analytics

#### Q21. Find the total count of customers in each `City`.

#### Q22. Calculate the average `TotalSpend` for each `FavCategory`.

```python
# Write your code here
FavCatdf = df.groupby("FavCategory")
FavCatdf["TotalSpend"].mean()
```

#### Q23. Find the maximum `TotalSpend` achieved in each `PreferredPayment` method group.

```python
# Write your code here
PPdf = df.groupby("PreferredPayment")
PPdf["TotalSpend"].max()
```

#### Q24. Determine which city has the highest overall average customer spending.

```python
# Write your code here
citydf = df.groupby("City")
print(f"{citydf["TotalSpend"].mean().idxmax()} : {citydf["TotalSpend"].mean().max()}")
```

#### Q25. Find all customers who have a `TotalSpend` higher than the overall average spend of the entire dataset.
